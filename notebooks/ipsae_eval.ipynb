```python
# Zip a folder for download (same approach as FreeBindCraft_UI)
import shutil
import subprocess
from pathlib import Path

import ipywidgets as widgets
from IPython.display import clear_output, display

ROOT = Path.cwd().resolve()
for base in [ROOT, *ROOT.parents, Path("/storage/viet/FreeBindCraft").resolve()]:
    if (base / "dependencies" / "IPSAE").exists():
        ROOT = base
        break


def _list_folders():
    return sorted(p.name for p in ROOT.iterdir() if p.is_dir() and not p.name.startswith(".")) or ["(no folders)"]


folder_dd = widgets.Dropdown(options=_list_folders(), description="Folder", layout=widgets.Layout(width="520px"))
refresh_btn = widgets.Button(description="Refresh", icon="refresh", layout=widgets.Layout(width="100px"))
zip_btn = widgets.Button(description="Zip folder", button_style="success", icon="file-archive")
status = widgets.HTML(value=f'<span style="color:#57606a">Select a folder under {ROOT}, then Zip.</span>')
log = widgets.Output()


def _refresh(_=None):
    folder_dd.options = _list_folders()


def _zip_folder(_):
    name = folder_dd.value
    src = ROOT / name
    with log:
        clear_output(wait=True)
        if not src.is_dir():
            status.value = '<span style="color:#cf222e">Folder not found.</span>'
            return
        if shutil.which("zip") is None:
            subprocess.run(["apt-get", "update"], check=False)
            subprocess.run(["apt-get", "install", "-y", "zip", "unzip"], check=True)
        # equivalent to: cd ROOT && zip -r <folder>.zip <folder>
        print(f"zip -r {name}.zip {name}")
        result = subprocess.run(["zip", "-r", f"{name}.zip", name], cwd=str(ROOT))
        if result.returncode == 0:
            status.value = f'<span style="color:#1a7f37;font-weight:600">Created {ROOT / (name + ".zip")}</span>'
        else:
            status.value = '<span style="color:#cf222e">zip failed (see log).</span>'


refresh_btn.on_click(_refresh)
zip_btn.on_click(_zip_folder)
display(widgets.VBox([widgets.HBox([folder_dd, refresh_btn]), zip_btn, status, log]))
```
